# Evaluación del impacto de estrategias bioclimáticas mediante simulación energética

## Descripción del proyecto

En el presente trabajo, se realizaron simulaciones a través del software EnergyPlus con la finalidad de definir el impacto de la aplicación de estrategias bioclimáticas en el tiempo en confort térmico, en una vivienda ubicada en Fresnillo, Zacatecas con 120 m2 de construcción, distribuidos en 3 recamaras, cocina, dos baños, sala y pasillo. Se evaluaron las condiciones internas térmicas de la vivienda construida, la cual presenta un comportamiento térmico hostil según la estación del año, temperaturainterna cálida en verano y temperatura fría en invierno. 

La aplicación de estrategias bioclimáticas mostró mejoras en las condiciones de confort, evidenciando el potencial de herramientas de simulación para optimizar el diseño de viviendas y edificios. Este trabajo busca impulsar el uso de software especializado en la evaluación de proyectos arquitectónicos y energéticos, promoviendo soluciones que aprovechen las características climáticas locales para mejorar la habitabilidad y reducir el consumo energético.

## Objetivo General

Evaluar el impacto de estrategias de arquitectura bioclimática sobre el confort térmico de una vivienda convencional ubicada en el municipio de Fresnillo, Zacatecas, mediante simulación energética utilizando SketchUp y EnergyPlus.

## Metodología

### 1.- Análisis bioclimático mediante Climate Consultant

Se utilizó el software Climate Consultant para determinar el rango de confort térmico exterior e identificar estrategias bioclimáticas adecuadas para el caso de estudio ubicado en Fresnillo, Zacatecas.

* Rango de confort térmico: Se determinó un rango de confort térmico entre 21 °C y 27 °C.

* Estrategias bioclimáticas seleccionadas

La principal estrategia identificada fue la GANANCIA SOLAR DIRECTA, complementada con medidas de CONTROL SOLAR.

Estrategias implementadas:
Ampliación de la ventana de la sala de 1.5 m² a 6.2 m² para incrementar la ganancia solar y el aprovechamiento de iluminación natural.
Incorporación de tragaluces en las tres recámaras con dimensiones de 0.5 m de altura y 1 m de ancho.
Instalación de protecciones solares en ventanas de las recámaras 2 y 3, con dimensiones de 1.74 m de ancho y longitud correspondiente al espacio de cada recámara.

Estas estrategias fueron implementadas con el objetivo de mejorar el confort térmico interior y reducir la incidencia de radiación solar directa durante el periodo de verano.



### 2.- Simulación energética mediante SketchUp y EnergyPlus

Se desarrolló un modelo energético de la vivienda utilizando SketchUp y EnergyPlus.

Se evaluaron dos escenarios:

Caso base sin implementación de estrategias bioclimáticas (diagnóstico).

### 2.- Calcular el porcentaje anual de horas dentro del rango de confort térmico por zona térmica para el caso base y para el escenario con estrategias bioclimáticas.



Para la zona de estudio, se considera una temperatura de confort (exterior) de 21-27 °C.
Para contabilizar el total de horas al año en confort térmico por zona térmica, se considerará el rango antes mencionado. 

```python
zonas = ["baño_completo", "baño", "cocina", "recamara_2", 
         "recamara_3", "recamara_principal", "pasillo_sala"]

# Crear tabla con conteo de horas en rango
tabla_confort = pd.DataFrame([
    {
        "Zona_térmica": zona,
        "Horas_en_confort_D": df[f"D_{zona}"].between(21, 27).sum(),
        "Horas_en_confort_E": df[f"E_{zona}"].between(21, 27).sum()
    }
    for zona in zonas
])

print(tabla_confort.to_string())
```

#### considerando que un año tiene 8,760 horas, se determinará el % de horas dentro del rango de confort y el porcentaje de incremento de confort posterior a la aplicación de Estrategias biocliomáticas.


```python
# considerando que un año tiene 8,760 horas, se determinará el % de horas dentro del rango de confort y el porcentaje de incremento de confort posterior a la aplicación de Estrategias biocliomáticas.
horas_año=8760
tabla_confort["horas_incremento"]=tabla_confort["Horas_en_confort_E"]-tabla_confort["Horas_en_confort_D"]
tabla_confort["%_horas_confort_D"]=(tabla_confort["Horas_en_confort_D"]/horas_año)*100
tabla_confort["%_horas_confort_E"]=(tabla_confort["Horas_en_confort_E"]/horas_año)*100
tabla_confort["%_horas_incremento"]=tabla_confort["%_horas_confort_E"]-tabla_confort["%_horas_confort_D"]
tabla_confort["promedio_horas_vivienda_D"]=tabla_confort["Horas_en_confort_D"].mean()
tabla_confort["promedio_horas_vivienda_E"]=tabla_confort["Horas_en_confort_E"].mean()
tabla_confort["promedio_horas_incremento_vivienda_E"]=tabla_confort["horas_incremento"].mean()
tabla_confort["%_horas_prom_confort_D"]=tabla_confort["%_horas_confort_D"].mean()
tabla_confort["%_horas_prom_confort_E"]=tabla_confort["%_horas_confort_E"].mean()
tabla_confort["promedio_%_incremento_confort_vivienda"]=tabla_confort["%_horas_incremento"].mean()
tabla_confort
```

Caso base (diagnóstico):
* todas las zonas térmicas tienen menos del 45% de horas al año dentro del rango de confort y el resto, más de 55% se encuentran fuera del rango de confort, yasea por encima de 27°C o por debajo de 21°C. 
* En promedio, para la vivienda, el 41.76% de las horas en un año, se encuentran dentro de la zona de confort.
* En promedio, para la vivienda, el 58.24% de las horas en un año, se encuentran fuera de la zona de confort.

Caso con Estrtategias bioclimáticas:

* En promedio, para la vivienda, el 47.76% de las horas en un año, se encuentran dentro de la zona de confort.
* En promedio, para la vivienda, el 58.24% de las horas en un año, se encuentran fuera de la zona de confort
* En la vivienda, se logró un incremento del 5.73% de horas en confort, lo que equivale a 503 horas.

```python
horas=tabla_confort.plot(kind="bar",
        x="Zona_térmica",
        y=["Horas_en_confort_D", "Horas_en_confort_E", "horas_incremento"], 
        xlabel="Zona térmica",
        ylabel="Horas",
        title="Horas en confort térmico al año",
)
for container in horas.containers:
    horas.bar_label(container, fmt="%.2f", label_type="edge", fontsize=8)
plt.legend(fontsize=7, loc="lower right", bbox_to_anchor=(1.15, 1))

plt.show()
plt.show()
                    
```

El efecto del rediseño de la ventana en la sala puede visualizarse en el incremento de horas en confort de esta área. De igual manera las 3 recamaras presentan un incremento de confort posterior a la construcción de tragaluces, favoreciendo la ganancia tanto de luz como de calor.

### 3.- Comparar el incremento del porcentaje de horas de confort térmico por zona térmica después de la implementación de estrategias bioclimáticas.

```python
porcentajes=tabla_confort.plot(kind="bar",
        x="Zona_térmica",
        y=["%_horas_confort_D", "%_horas_confort_E", "%_horas_incremento"], 
        xlabel="Zona térmica",
        ylabel="Horas",
        title="% de horas en confort térmico al año",
)
for container in porcentajes.containers:
    porcentajes.bar_label(container, fmt="%.2f", label_type="edge", fontsize=8)

plt.legend(fontsize=7, loc="lower right", bbox_to_anchor=(1.15, 1))
plt.show()
```

#### 4.3 determinar temperatura promedio, máxima y mínima mensual por ZT

```python
def plot_zona(df, zona, fecha_inicio, fecha_fin):
    """
    df: DataFrame con datos
    zona: nombre base de la zona (ej. 'recamara_principal', 'recamara_2', 'cocina')
    fecha_inicio, fecha_fin: rango de fechas a filtrar
    """
    # Filtrar fechas
    df_filtrado = df[(df['date'] >= fecha_inicio) & (df['date'] <= fecha_fin)]

    # Columnas a graficar
    cols = [f"D_{zona}", f"E_{zona}"]

    # Graficar
    ax = df_filtrado.plot(
        kind="line",
        x="date",
        y=cols,
        xlabel="días",
        ylabel="Temperatura °C",
        figsize=(12,6),
        marker="o"
    )

    # Calcular y dibujar líneas horizontales con etiquetas
    for col, color in zip(cols, ["red", "blue"]):
        max_val = df_filtrado[col].max()
        min_val = df_filtrado[col].min()
        prom_val = df_filtrado[col].mean()

        plt.axhline(y=max_val, color=color, linestyle="--", label=f"Máx {col}: {max_val:.1f}°C")
        ax.text(df_filtrado["date"].iloc[-1], max_val, f"{max_val:.1f}°C", color=color, va="bottom")

        plt.axhline(y=min_val, color=color, linestyle="--", label=f"Mín {col}: {min_val:.1f}°C")
        ax.text(df_filtrado["date"].iloc[-1], min_val, f"{min_val:.1f}°C", color=color, va="top")

        plt.axhline(y=prom_val, color="black" if "E_" in col else "gray", linestyle="--", 
                    label=f"Prom {col}: {prom_val:.1f}°C")
        ax.text(df_filtrado["date"].iloc[-1], prom_val, f"{prom_val:.1f}°C", 
                color="black" if "E_" in col else "gray", va="bottom")

    plt.legend(fontsize=6.5, loc="best")
    plt.tight_layout()
    plt.show()

```

PRUEBAS AB

Ho: El promedio de la temperatura anual del diagnóstico y de la aplicación de estrategias bioclimáticas es la misma.

H alternativa: El promedio de la temperatura anual del diagnóstico y de la aplicación de estrategias bioclimáticas no es la misma.

```python
alpha = 0.05

# Extraer las puntuaciones individuales
D_prom_t = df["D_recamara_2"]
E_prom_t = df["E_recamara_2"]

# Prueba t de muestras independientes

results = stats.ttest_rel(D_prom_t, E_prom_t)

print("Valor p:", results.pvalue)

if results.pvalue < alpha:
    print("Rechazamos H0: hay diferencia significativa")
else:
    print("No rechazamos H0: no hay evidencia de diferencia")
```

las estrategias aplicadas en la recámara_2 generaron un cambio estadísticamente significativo en la temperatura interna, lo cual respalda que las estrategias tiene un efecto real.

```python

alpha = 0.05

# Series completas
D_vals = df["D_recamara_2"]
E_vals = df["E_recamara_2"]

# Prueba t pareada
results = stats.ttest_rel(E_vals,D_vals,)

print("Valor p:", results.pvalue)

# Calcular Cohen's d para muestras relacionadas
diff = E_vals - D_vals 
cohens_d = diff.mean() / diff.std(ddof=1)

print("Cohen's d:", cohens_d)

if results.pvalue < alpha:
    print("Rechazamos H0: diferencia significativa")
else:
    print("No rechazamos H0: no hay evidencia de diferencia")
```

Las estrategias bioclimáticas aplicadas en la recámara 2 lograron aumentar la temperatura interna de manera significativa y con un efecto muy grande, lo cual cumple tu objetivo de mejorar el confort.
